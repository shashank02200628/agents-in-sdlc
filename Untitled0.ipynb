```python
import os
import json

os.makedirs("portify/data", exist_ok=True)
os.makedirs("portify/output", exist_ok=True)
os.makedirs("portify/logs", exist_ok=True)

print("===== PORTIFY: STUDENT PORTFOLIO BUILDER =====")

name = input("Enter your full name: ")
college = input("Enter your college name: ")
department = input("Enter your department: ")
year = input("Enter your current year/semester: ")
skills = input("Enter your skills (comma-separated): ")
github_username = input("Enter your GitHub username (or leave blank): ")

student = {
    "name": name,
    "college": college,
    "department": department,
    "year": year,
    "skills": [s.strip() for s in skills.split(",") if s.strip()],
    "github_username": github_username.strip()
}

with open("portify/data/student.json", "w") as f:
    json.dump(student, f, indent=4)

print("\nStudent profile saved successfully!")
```

```python
import requests
import json
import os

student_file = "portify/data/student.json"

with open(student_file, "r") as f:
    student = json.load(f)

username = student["github_username"]
repositories = []

if username:
    # Validate the username before inserting it into the API URL
    import re

    if not re.fullmatch(r"[A-Za-z0-9-]{1,39}", username):
        print("Invalid GitHub username format.")
    else:
        api_url = f"https://api.github.com/users/{username}/repos"
        params = {
            "per_page": 100,
            "sort": "updated"
        }

        try:
            response = requests.get(api_url, params=params, timeout=15)

            if response.status_code == 200:
                repositories = response.json()

                if isinstance(repositories, list):
                    repositories = [
                        {
                            "name": repo.get("name", ""),
                            "description": repo.get("description") or "No description available.",
                            "language": repo.get("language") or "Not specified",
                            "url": repo.get("html_url", "")
                        }
                        for repo in repositories
                    ]
                    print(f"Successfully fetched {len(repositories)} public repositories!")
                else:
                    repositories = []
                    print("Unexpected response received from GitHub.")
            elif response.status_code == 404:
                print("GitHub user not found. Check the username.")
            else:
                print("GitHub API returned status:", response.status_code)

        except requests.RequestException as error:
            print("Unable to connect to GitHub:", error)
else:
    print("No GitHub username entered. You can still create your portfolio.")

with open("portify/data/repositories.json", "w") as f:
    json.dump(repositories, f, indent=4)

print("Repository data saved.")
```

```python
import json
import html
import os
import re

with open("portify/data/student.json", "r") as f:
    student = json.load(f)

with open("portify/data/repositories.json", "r") as f:
    repositories = json.load(f)

def safe(value):
    return html.escape(str(value), quote=True)

skills_html = "".join(
    f'<span class="skill">{safe(skill)}</span>'
    for skill in student["skills"]
)

repo_cards = ""

for repo in repositories:
    repo_url = repo.get("url", "")

    # Only allow links to GitHub repository pages
    if not re.fullmatch(
        r"https://github\.com/[A-Za-z0-9_.-]+/[A-Za-z0-9_.-]+",
        repo_url
    ):
        repo_url = ""

    link_html = (
        f'<a href="{safe(repo_url)}" target="_blank" rel="noopener noreferrer">View Repository ↗</a>'
        if repo_url else ""
    )

    repo_cards += f"""
    <div class="card">
        <h3>{safe(repo.get("name", "Project"))}</h3>
        <p>{safe(repo.get("description", "No description available."))}</p>
        <p><strong>Language:</strong> {safe(repo.get("language", "Not specified"))}</p>
        {link_html}
    </div>
    """

if not repo_cards:
    repo_cards = """
    <div class="card">
        <h3>My Projects</h3>
        <p>My projects will be added here soon.</p>
    </div>
    """

portfolio_html = f"""<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<title>{safe(student["name"])} | Portfolio</title>
<style>
* {{ box-sizing: border-box; }}
body {{
    margin: 0;
    font-family: Arial, sans-serif;
    background: #f4f6fb;
    color: #20243a;
    line-height: 1.6;
}}
header {{
    background: linear-gradient(135deg, #141e30, #334e8c);
    color: white;
    text-align: center;
    padding: 55px 20px;
}}
header h1 {{ font-size: 2.5rem; margin-bottom: 8px; }}
section {{ max-width: 900px; margin: 30px auto; padding: 0 20px; }}
.panel {{
    background: white;
    padding: 25px;
    border-radius: 14px;
    box-shadow: 0 4px 18px #0000000b;
}}
.skills {{ display: flex; flex-wrap: wrap; gap: 10px; }}
.skill {{
    background: #e4eaff;
    color: #233c8b;
    padding: 7px 13px;
    border-radius: 20px;
}}
.projects {{
    display: grid;
    grid-template-columns: repeat(auto-fit, minmax(230px, 1fr));
    gap: 16px;
}}
.card {{
    background: white;
    padding: 20px;
    border-radius: 12px;
    border: 1px solid #e1e5f0;
    overflow-wrap: anywhere;
}}
.card a {{ color: #3158b7; text-decoration: none; font-weight: bold; }}
footer {{ text-align: center; padding: 25px; color: #68708a; }}
</style>
</head>
<body>
<header>
    <h1>{safe(student["name"])}</h1>
    <p>{safe(student["department"])} | {safe(student["year"])}</p>
    <p>{safe(student["college"])}</p>
</header>

<section>
    <div class="panel">
        <h2>About Me</h2>
        <p>Hello! I am a student interested in technology, programming, and software development.</p>
    </div>
</section>

<section>
    <div class="panel">
        <h2>Technical Skills</h2>
        <div class="skills">{skills_html or "<p>Skills will be added soon.</p>"}</div>
    </div>
</section>

<section>
    <h2>My GitHub Projects</h2>
    <div class="projects">{repo_cards}</div>
</section>

<footer>Generated with Portify — Student Portfolio Builder</footer>
</body>
</html>
"""

os.makedirs("portify/output", exist_ok=True)

output_path = "portify/output/index.html"

with open(output_path, "w", encoding="utf-8") as f:
    f.write(portfolio_html)

print("Portfolio generated successfully!")
print("File location:", output_path)
```
